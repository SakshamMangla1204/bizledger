```python
def net_income(revenue,expense):
    income=revenue-expense

    if income > 0:
        status="Profit"

    elif income <0 :
            status = "Loss"

    else:

      status="break even "


      return income,status

   

```

```python
def revenue(income,expense):
    if income==0:
        return None

    revenue=income-expense
    return revenue

```

```python
def profit_margin(income,revenue):
    if revenue==0:
        return None
    profit_margin=(income/revenue)*100

    return profit_margin
```

```python
def expense_ratio(expense,revenue):
    if revenue==0:
        return None 

    expense_ratio=(expense/revenue)*100
    return expense_ratio
```

```python
def current_ratio(current_assets,current_liabilities):
    if current_liabilities==0:
        return None

    current_ratio=(current_assets/current_liabilities)
    return current_ratio
```
